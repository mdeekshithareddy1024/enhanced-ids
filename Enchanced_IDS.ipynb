```python
import pandas as pd

# Load datasets correctly
train_data = pd.read_csv("/content/KDDTrain+ (1).csv", header=None)
test_data = pd.read_csv("/content/KDDTest+ (1).csv", header=None)

print("Train shape:", train_data.shape)
print("Test shape:", test_data.shape)

# Separate features, labels and difficulty
X_train = train_data.iloc[:, :-2].copy()
y_train = train_data.iloc[:, -2].copy()

X_test = test_data.iloc[:, :-2].copy()
y_test = test_data.iloc[:, -2].copy()

# Give correct column names
column_names = [
    "duration", "protocol_type", "service", "flag",
    "src_bytes", "dst_bytes", "land", "wrong_fragment",
    "urgent", "hot", "num_failed_logins", "logged_in",
    "num_compromised", "root_shell", "su_attempted",
    "num_root", "num_file_creations", "num_shells",
    "num_access_files", "num_outbound_cmds", "is_host_login",
    "is_guest_login", "count", "srv_count", "serror_rate",
    "srv_serror_rate", "rerror_rate", "srv_rerror_rate",
    "same_srv_rate", "diff_srv_rate", "srv_diff_host_rate",
    "dst_host_count", "dst_host_srv_count",
    "dst_host_same_srv_rate", "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate", "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate", "dst_host_srv_serror_rate",
    "dst_host_rerror_rate", "dst_host_srv_rerror_rate"
]

X_train.columns = column_names
X_test.columns = column_names

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

display(X_train.head())
```

```python
import numpy as np

# Convert data to numeric arrays
X_train_np = X_train.astype(float).values
X_test_np = X_test.astype(float).values

# Calculate mean and range using TRAINING data only
mean = np.mean(X_train_np, axis=0)
feature_range = np.max(X_train_np, axis=0) - np.min(X_train_np, axis=0)

# Avoid division by zero
feature_range[feature_range == 0] = 1

# Mean normalization
X_train_normalized = (X_train_np - mean) / feature_range
X_test_normalized = (X_test_np - mean) / feature_range

print("Normalization completed!")

print("X_train normalized shape:", X_train_normalized.shape)
print("X_test normalized shape:", X_test_normalized.shape)

print("\nFirst 5 normalized training rows:")
print(X_train_normalized[:5])
```

```python
# Convert labels into 2 classes
# Normal = 0
# Attack = 1

y_train_binary = (y_train != "normal").astype(int).values
y_test_binary = (y_test != "normal").astype(int).values

print("Training labels:")
print(pd.Series(y_train_binary).value_counts())

print("\nTesting labels:")
print(pd.Series(y_test_binary).value_counts())
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Use the normalized training data
X_mvo = X_train_normalized
y_mvo = y_train_binary

# Take a smaller sample so MVO runs faster in Colab
sample_size = 5000

np.random.seed(42)

sample_indices = np.random.choice(
    len(X_mvo),
    size=sample_size,
    replace=False
)

X_sample = X_mvo[sample_indices]
y_sample = y_mvo[sample_indices]

# Split sample into training and validation
X_fit, X_val, y_fit, y_val = train_test_split(
    X_sample,
    y_sample,
    test_size=0.2,
    random_state=42,
    stratify=y_sample
)

print("MVO sample shape:", X_sample.shape)
print("MVO training shape:", X_fit.shape)
print("MVO validation shape:", X_val.shape)
```

```python
# ==========================================
# MVO FEATURE SELECTION
# ==========================================

import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Total number of features
n_features = X_fit.shape[1]

# Number of features we want to select
target_features = 24

# MVO parameters
n_universes = 15
max_iterations = 10

np.random.seed(42)


# ------------------------------------------
# Create a universe with exactly 24 features
# ------------------------------------------

def create_universe():

    universe = np.zeros(n_features, dtype=int)

    selected = np.random.choice(
        n_features,
        target_features,
        replace=False
    )

    universe[selected] = 1

    return universe


# ------------------------------------------
# Keep every universe at exactly 24 features
# ------------------------------------------

def repair_universe(universe):

    selected = np.where(universe == 1)[0]
    not_selected = np.where(universe == 0)[0]

    # If more than 24 features
    if len(selected) > target_features:

        remove = np.random.choice(
            selected,
            len(selected) - target_features,
            replace=False
        )

        universe[remove] = 0

    # If fewer than 24 features
    elif len(selected) < target_features:

        add = np.random.choice(
            not_selected,
            target_features - len(selected),
            replace=False
        )

        universe[add] = 1

    return universe


# ------------------------------------------
# Initial population
# ------------------------------------------

population = np.array([
    create_universe()
    for _ in range(n_universes)
])


# ------------------------------------------
# Fitness function
# ------------------------------------------

def calculate_fitness(universe):

    selected_features = np.where(universe == 1)[0]

    model = DecisionTreeClassifier(
        max_depth=6,
        random_state=42
    )

    model.fit(
        X_fit[:, selected_features],
        y_fit
    )

    predictions = model.predict(
        X_val[:, selected_features]
    )

    accuracy = accuracy_score(
        y_val,
        predictions
    )

    return accuracy


# ------------------------------------------
# MVO PROCESS
# ------------------------------------------

best_universe = None
best_fitness = -1


for iteration in range(max_iterations):

    # Calculate fitness
    fitness_values = np.array([
        calculate_fitness(universe)
        for universe in population
    ])


    # Find current best universe
    current_best_index = np.argmax(fitness_values)

    current_best_fitness = fitness_values[
        current_best_index
    ]


    # Update global best
    if current_best_fitness > best_fitness:

        best_fitness = current_best_fitness

        best_universe = population[
            current_best_index
        ].copy()


    print(
        "Iteration:",
        iteration + 1,
        "| Best Accuracy:",
        round(best_fitness, 4)
    )


    # --------------------------------------
    # Sort universes
    # --------------------------------------

    sorted_indices = np.argsort(
        fitness_values
    )[::-1]

    population = population[
        sorted_indices
    ]


    # --------------------------------------
    # White-hole / Black-hole operation
    # --------------------------------------

    for i in range(n_universes):

        for j in range(n_features):

            transfer_probability = (
                n_universes - i
            ) / n_universes


            if np.random.random() < transfer_probability:

                source_index = np.random.randint(
                    0,
                    n_universes
                )

                population[i, j] = population[
                    source_index, j
                ]


    # --------------------------------------
    # Wormhole operation
    # --------------------------------------

    wormhole_probability = (
        0.2
        + 0.8 *
        ((iteration + 1) / max_iterations)
    )


    for i in range(n_universes):

        for j in range(n_features):

            if np.random.random() < wormhole_probability:

                population[i, j] = (
                    best_universe[j]
                )


    # --------------------------------------
    # IMPORTANT:
    # Keep exactly 24 features
    # --------------------------------------

    for i in range(n_universes):

        population[i] = repair_universe(
            population[i]
        )


# ==========================================
# FINAL RESULT
# ==========================================

selected_indices = np.where(
    best_universe == 1
)[0]

# Safety check
if len(selected_indices) != 24:

    best_universe = repair_universe(
        best_universe
    )

    selected_indices = np.where(
        best_universe == 1
    )[0]


print("\n==============================")
print("MVO FEATURE SELECTION COMPLETE")
print("==============================")

print(
    "Best validation accuracy:",
    round(best_fitness, 4)
)

print(
    "Number of selected features:",
    len(selected_indices)
)

print("\nSelected feature indexes:")
print(selected_indices)
```

```python
# ==========================================
# APPLY MVO SELECTED FEATURES
# ==========================================

# Use the 24 features selected by MVO
X_train_selected = X_train_normalized[:, selected_indices]
X_test_selected = X_test_normalized[:, selected_indices]

print("Original training shape:")
print(X_train_normalized.shape)

print("\nSelected training shape:")
print(X_train_selected.shape)

print("\nOriginal testing shape:")
print(X_test_normalized.shape)

print("\nSelected testing shape:")
print(X_test_selected.shape)
```

```python
# ==========================================
# PREPARE DATA FOR CNN-BiGRU
# ==========================================

# CNN-BiGRU expects 3D input:
# (samples, features, channels)

X_train_cnn = X_train_selected.reshape(
    X_train_selected.shape[0],
    X_train_selected.shape[1],
    1
)

X_test_cnn = X_test_selected.reshape(
    X_test_selected.shape[0],
    X_test_selected.shape[1],
    1
)

print("CNN training shape:", X_train_cnn.shape)
print("CNN testing shape:", X_test_cnn.shape)

print("Training labels shape:", y_train_binary.shape)
print("Testing labels shape:", y_test_binary.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

print("TensorFlow version:", tf.__version__)
```

```python
# CNN layer

inputs = layers.Input(
    shape=(24, 1)
)

cnn = layers.Conv1D(
    filters=64,
    kernel_size=3,
    activation="relu",
    padding="same"
)(inputs)

cnn = layers.MaxPooling1D(
    pool_size=2
)(cnn)

print("CNN layer created successfully!")
```

```python
# BiGRU layer

bigru = layers.Bidirectional(
    layers.GRU(
        64,
        return_sequences=True
    )
)(cnn)

print("BiGRU layer created successfully!")
```

```python
# Attention layer

attention = layers.Attention()(
    [bigru, bigru]
)

print("Attention layer created successfully!")
```

```python
# ==========================================
# FINAL OUTPUT LAYER
# ==========================================

# Convert the Attention sequence into one vector
flatten = layers.Flatten()(attention)

# Fully connected layer
dense = layers.Dense(
    64,
    activation="relu"
)(flatten)

# Binary classification output
outputs = layers.Dense(
    1,
    activation="sigmoid"
)(dense)

# Create the complete model
model = Model(
    inputs=inputs,
    outputs=outputs
)

print("CNN-BiGRU-Attention model created successfully!")
```

```python
# ==========================================
# COMPILE THE MODEL
# ==========================================

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
# ==========================================
# MODEL SUMMARY
# ==========================================

model.summary()
```

```python
# ==========================================
# TRAIN THE CNN-BiGRU-ATTENTION MODEL
# ==========================================

history = model.fit(
    X_train_cnn,
    y_train_binary,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# ALO Hyperparameter Search Space

search_space = {
    "learning_rate": [0.0001, 0.0005, 0.001],
    "gru_units": [32, 64, 128],
    "dense_units": [32, 64, 128],
    "batch_size": [64, 128, 256]
}

print("ALO hyperparameter search space:")
for parameter, values in search_space.items():
    print(parameter, ":", values)
```

```python
import random
import tensorflow as tf
from tensorflow.keras import layers, Model

def build_model(learning_rate, gru_units, dense_units):

    inputs = layers.Input(shape=(24, 1))

    # CNN
    cnn = layers.Conv1D(
        filters=64,
        kernel_size=3,
        activation="relu",
        padding="same"
    )(inputs)

    cnn = layers.MaxPooling1D(pool_size=2)(cnn)

    # BiGRU
    bigru = layers.Bidirectional(
        layers.GRU(gru_units, return_sequences=True)
    )(cnn)

    # Attention
    attention = layers.Attention()([bigru, bigru])

    # Dense
    flatten = layers.Flatten()(attention)

    dense = layers.Dense(
        dense_units,
        activation="relu"
    )(flatten)

    outputs = layers.Dense(
        1,
        activation="sigmoid"
    )(dense)

    model = Model(inputs=inputs, outputs=outputs)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


print("Model-building function created successfully!")
```

```python
import numpy as np
import random

# Number of ALO agents and iterations
num_agents = 3
num_iterations = 3

best_accuracy = 0
best_params = None

for iteration in range(num_iterations):

    print(f"\nALO Iteration {iteration + 1}/{num_iterations}")

    for agent in range(num_agents):

        learning_rate = random.choice(search_space["learning_rate"])
        gru_units = random.choice(search_space["gru_units"])
        dense_units = random.choice(search_space["dense_units"])
        batch_size = random.choice(search_space["batch_size"])

        print(
            f"Agent {agent + 1}: "
            f"LR={learning_rate}, "
            f"GRU={gru_units}, "
            f"Dense={dense_units}, "
            f"Batch={batch_size}"
        )

        temp_model = build_model(
            learning_rate,
            gru_units,
            dense_units
        )

        history_temp = temp_model.fit(
            X_train_cnn,
            y_train_binary,
            epochs=3,
            batch_size=batch_size,
            validation_split=0.2,
            verbose=0
        )

        val_accuracy = max(history_temp.history["val_accuracy"])

        print(f"Validation Accuracy: {val_accuracy:.4f}")

        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy

            best_params = {
                "learning_rate": learning_rate,
                "gru_units": gru_units,
                "dense_units": dense_units,
                "batch_size": batch_size
            }

            print("★ New best parameters found!")

print("\n==============================")
print("ALO OPTIMIZATION COMPLETE")
print("==============================")

print("Best Validation Accuracy:", best_accuracy)
print("Best Parameters:")

for key, value in best_params.items():
    print(f"{key}: {value}")
```

```python
import random

num_agents = 2
num_iterations = 2

best_accuracy = 0
best_params = None

for iteration in range(num_iterations):

    print(f"\nALO Iteration {iteration + 1}/{num_iterations}")

    for agent in range(num_agents):

        learning_rate = random.choice(search_space["learning_rate"])
        gru_units = random.choice(search_space["gru_units"])
        dense_units = random.choice(search_space["dense_units"])
        batch_size = random.choice(search_space["batch_size"])

        print(
            f"Agent {agent + 1}: "
            f"LR={learning_rate}, "
            f"GRU={gru_units}, "
            f"Dense={dense_units}, "
            f"Batch={batch_size}"
        )

        temp_model = build_model(
            learning_rate,
            gru_units,
            dense_units
        )

        history_temp = temp_model.fit(
            X_alo,
            y_alo,
            epochs=2,
            batch_size=batch_size,
            validation_split=0.2,
            verbose=0
        )

        val_accuracy = max(history_temp.history["val_accuracy"])

        print(f"Validation Accuracy: {val_accuracy:.4f}")

        if val_accuracy > best_accuracy:
            best_accuracy = val_accuracy

            best_params = {
                "learning_rate": learning_rate,
                "gru_units": gru_units,
                "dense_units": dense_units,
                "batch_size": batch_size
            }

            print("★ New best parameters found!")

        del temp_model

print("\n==============================")
print("FAST ALO SEARCH COMPLETE")
print("==============================")

print("Best Validation Accuracy:", best_accuracy)
print("Best Parameters:")

for key, value in best_params.items():
    print(f"{key}: {value}")
```
